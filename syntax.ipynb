# naming convention in python 
"""- variable name should start with a letter or an underscore
- variable name can only contain alphanumeric characters and underscores
- variable name cannot be a reserved keyword
- variable name should be descriptive and meaningful
- variable name should not be too long or too short
- variable name should follow the snake_case convention (lowercase letters and underscores) 
- case sensitive (age, Age and AGE are three different variables)"""

# Operators in python

# Arithmetic operators

```python

a=10    
b=3
print(a+b)  # addition
print(a-b)  # subtraction
print(a*b)  # multiplication
print(a/b)  # division
print(a//b) # floor division
print(a%b)  # modulus   
print(a**b) # exponentiation
print(-a)   # negation
print(+a)   # positive

```

# comparison operators

```python
# comparison operators
a=10
b=20
print(a==b)  # equal to
print(a!=b)  # not equal to
print(a>b)   # greater than
print(a<b)   # less than
print(a>=b)  # greater than or equal to
print(a<=b)  # less than or equal to

```

# identity operator

```python
print(a is b)
```

```python
print(a is not b)
```

# membership operator

```python
print(a in [10,20,30]) 
```

```python
print(a not in [10,20,30]) 
```

# logical operators


```python
a=10
b=20    
print(a>5 and b>15)  # logical AND
print(a>5 or b>25)   # logical OR
print(not(a>5))     # logical NOT
print(not(a>15))     # logical NOT
print(not(a>5 and b>15))  # logical NOT of AND
print(not(a>5 or b>25))   # logical NOT of OR
print((a>5 and b>15) or (a>5 or b>25))  # combination of logical operators
print((a>5 and b>15) and (a>5 or b>25))  # combination of logical operators
print(not((a>5 and b>15) or (a>5 or b>25)))  # logical NOT of combination
print(not((a>5 and b>15) and (a>5 or b>25)))  # logical NOT of combination
```
