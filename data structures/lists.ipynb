<h1><b> List </b></h1>

A List in Python is a built-in data structure used to store multiple items in a single variable. 
- Lists are ordered, meaning elements maintain their position, and they are mutable, so items can be added, removed, or modified after creation. 
- A list can contain duplicate values and can also store different types of data such as integers, strings, floats, and Boolean values together.
-  Lists are created using square brackets [], with elements separated by commas. 
- They support indexing, slicing, looping, and many useful methods like append(), insert(), remove(), and sort(). Because of their flexibility and ease of use, lists are one of the most commonly used data structures in Python programming.

```python
# creating list 
list1=["apple", "banana", "cherry",10,20,30]
print(list1)
print(type(list1))
```

```python
# accessing list items
print(list1[0]) # first item
print(list1[3]) # fourth item   \
print(list1[-1]) # last item
print(list1[0:3]) # first three items
print(list1[3:]) # from fourth item to end
print(list1[:3]) # from start to third item
print(list1[-3:]) # last three items
print(list1[::])
```

# modifying list items

```python
# 1. Changing Elements

fruits = ["apple", "banana", "orange"]

fruits[1] = "mango"

print(fruits)
```

```python
# 2. Modifying Multiple Elements Using Slicing

numbers = [1, 2, 3, 4, 5]

numbers[1:4] = [20, 30, 40]

print(numbers)
```

# List methods

```python
# append() method to add an element to the end of the list
nums = [1, 2, 3]

nums.append(4)

print(nums)



```

```python
# insert() method to add an element at a specific index list.insert(index, value)
nums = [1, 3]

nums.insert(1, 2)

print(nums)

```

```python
#extend() method to add elements from another list to the end of the current list

list1 = [1, 2, 3]
list2 = [4, 5, 6]
list1.extend(list2)
print(list1)
```

```python
# remove() method to remove the first occurrence of a specified value from the list
fruits = ["apple", "banana", "cherry", "banana"]
fruits.remove("banana")
print(fruits)
```

```python
# 5. pop() method to remove an element at a specific index and return it
numbers = [10, 20, 30, 40, 50]
removed_element = numbers.pop(2)
print("Removed element:", removed_element)
print("Updated list:", numbers)
```

```python
# pop() Without Index
numbers = [10, 20, 30, 40, 50]
removed_element = numbers.pop()

print("Removed element:", removed_element)
print("Updated list:", numbers)
```

```python
# 6. clear() method to remove all elements from the list
my_list = [1, 2, 3, 4, 5]   
my_list.clear()
print(my_list)
```

```python
# 7. index() method to find the index of the first occurrence of a specified value in the list
my_list = [10, 20, 30, 40, 50]
index = my_list.index(30)
print("Index of 30:", index)
```

```python
# count() method to count the number of occurrences of a specified value in the list
my_list = [1, 2, 3, 2, 4, 2]
count = my_list.count(2)
print("Count of 2:", count)
```

```python
# sort() method to sort the elements of the list in ascending order
numbers = [5, 2, 9, 1, 5, 6]
numbers.sort()
print(numbers)
```

```python
# Descending Order
numbers = [5, 2, 9, 1, 5, 6]
numbers.sort(reverse=True)  
print(numbers)
```

```python
# reverse() method to reverse the order of the elements in the list
my_list = [1, 2, 3, 4, 5]
my_list.reverse()
print(my_list)
```

```python
# copy() method to create a shallow copy of the list
my_list = [1, 2, 3, 4, 5]
my_list_copy = my_list.copy()
print(my_list_copy)
```

```python
# append() vs extend()
list1 = [1, 2, 3]
list2 = [4, 5, 6]
list1.append(list2)
print("After append:", list1)

a = [1, 2]

a.extend([3, 4])

print("after extend:", a)
```

```python
# slicing to modify list
my_list = [1, 2, 3, 4, 5]
my_list[1:4] = [20, 30, 40]
print(my_list)
print(my_list[1:4])
print(my_list[::1])
print(my_list[::2])
print(my_list[::-1])
print(my_list[-1:4])
print(my_list[-4:-1])
print(my_list[-4:4])
```

# Iterating over list

```python
fruits = ["apple", "banana", "orange"]
```

```python
# 1. Using for Loop (Most Common)
for fruit in fruits:
    print(fruit)
```

```python
# 2. Using for Loop with Index
for i in range(len(fruits)):
    print(fruits[i])
```

```python
# 3. Using while Loop
i = 0
while i < len(fruits):
    print(fruits[i])
    i += 1
```

```python
# 4. Using enumerate()

# Best way when you need both index and value.

for index, fruit in enumerate(fruits):
    print(index, fruit)
```

```python
# 6. Iterating in Reverse Order
# reversed function returns an iterator that accesses the given sequence in the reverse order.
for fruit in reversed(fruits):
    print(fruit)
```

```python
# 7. Using Negative Indexing
for i in range(-1, -len(fruits)-1, -1):
    print(fruits[i])
```

# LIST COMPREHENSION

List comprehension is a short and powerful way to create lists.

```python
# 1. Simple List Comprehension
squares = [x**2 for x in range(1, 6)]
print(squares)
```

```python
# 2. List Comprehension with Condition
even_squares = [x**2 for x in range(1, 11) if x % 2 == 0]
print(even_squares)
```

```python
# 3. if-else in List Comprehension
numbers = [1, 2, 3, 4, 5]
labels = ["Even" if x % 2 == 0 else "Odd" for x in numbers]
print(labels)
```

```python
# 4. String Manipulation
fruits = ["apple", "banana", "cherry"]
capitalized_fruits = [fruit.upper() for fruit in fruits]
print(capitalized_fruits)
```

```python
# 5. Creating List from Existing List
original_list = [1, 2, 3, 4, 5]
squared_list = [x**2 for x in original_list]
print(squared_list)
```

```python
# 6. Nested Loop List Comprehension

matrix = [[1, 2, 3], [4, 5, 6], [7, 8, 9]]
flattened = [num for row in matrix for num in row]
print(flattened)
```

```python
# 9. Nested List Comprehension
matrix = [[1, 2, 3], [4, 5, 6], [7, 8, 9]]
squared_matrix = [[num**2 for num in row] for row in matrix]
print(squared_matrix)
```
